```python
import pandas as pd

# 1. Load raw FBref export
df = pd.read_csv("ocsc_2026_standard.csv")

# 2. Remove FBref summary rows (Squad Total / Opponent Total)
players = df[df["Pos"].notna()].copy()

# 3. Clean numeric columns
players["Min"] = pd.to_numeric(players["Min"].astype(str).str.replace(",", ""), errors="coerce")
players["Gls"] = pd.to_numeric(players["Gls"], errors="coerce")
players["Ast"] = pd.to_numeric(players["Ast"], errors="coerce")

# 4. Create per-90 efficiency metrics
players["G_plus_A"] = players["Gls"] + players["Ast"]
players["Goals_per_90"] = ((players["Gls"] / players["Min"]) * 90).round(2)
players["GA_per_90"] = ((players["G_plus_A"] / players["Min"]) * 90).round(2)

# 5. Filter for 600+ minutes and sort by Goals_per_90
leaderboard = (
    players[players["Min"] >= 600]
    .sort_values(by="Goals_per_90", ascending=False)
    [["Player", "Pos", "Min", "Gls", "Ast", "G_plus_A", "Goals_per_90", "GA_per_90"]]
)

# 6. Export the clean leaderboard to a new CSV (index=False removes row numbers)
leaderboard.to_csv("ocsc_2026_top_scorers.csv", index=False)

print("Exported ocsc_2026_top_scorers.csv successfully!")
leaderboard.head(10)
```

**Source:** FBref (USL Championship)

### What This Script Does
* Removes FBref `Squad Total` and `Opponent Total` summary rows via (`Pos.notna()`).
* Converts comma-formatted `Min` strings into numeric values.
* Calculates `G+A`, `Goals_per_90`, and `GA_per_90` for players with **600+ minutes**.
* Exports into a clean, analysis-ready CSV (`ocsc_2026_top_scorers.csv`).
